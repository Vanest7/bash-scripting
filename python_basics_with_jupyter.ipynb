## Exercise 1: Balance and profit

Calculate the net profit and return percentage of a simulated session.

```python
initial_balance = 80
final_balance = 80
strategy = "martingala"
```

```python
profit = final_balance - initial_balance
profit_percentage = profit / initial_balance * 100
has_profit = profit > 0

print("Net profit:", profit)
print("Return percentage:", profit_percentage)
print("Profitable session:", has_profit)
```

```python
if profit > 0:
    print("The session ended with a profit.")
elif profit < 0:
    print("The session ended with a loss.")
else:
    print("The session ended with no profit or loss.")
```

```python
session_profits = [20, 0, -20, 10]

for session_profit in session_profits:
    print("Session profit:", session_profit)
```

```python
winning_sessions = 0

for session_profit in session_profits:
    if session_profit > 0:
        winning_sessions += 1

print("Winning sessions:", winning_sessions)
```

```python
def calculate_profit(initial_balance, final_balance):
    return final_balance - initial_balance
```

```python
profit = calculate_profit(80, 65)

print("Net profit:", profit)
```

```python
def calculate_return_percentage(initial_balance, final_balance):
    return (final_balance - initial_balance) / initial_balance *100

percentage = calculate_return_percentage(80, 100)
print(percentage)
```
